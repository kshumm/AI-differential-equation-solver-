```python
# Install packages (run this only once in Colab)
# !pip install sympy google-generativeai

import sympy as sp
import google.generativeai as genai

# 🔑 STEP 1: Put your API key here (inside quotes)
#genai.configure(api_key="AIzaSyAWnvPakZaPerfPXPvM1Q7FqZ1dIsIBRv0")

# STEP 2: Define variables
x = sp.symbols('x')
y = sp.Function('y')

print("=== AI Differential Equation Solver ===")

# STEP 3: Example Differential Equation
eq_input = "diff(y(x), x, 2) - diff(y(x), x) + y(x)"

print("\nEquation:")
print(eq_input)

try:
    # STEP 4: Solve using SymPy
    eq = sp.sympify(eq_input)
    solution = sp.dsolve(eq, y(x))

    print("\n✅ Solution:")
    print(solution)

    # Detect order
    order = sp.ode_order(eq, y(x))
    print(f"Step 1: This is a {order}-order differential equation.")

    # General explanation
    if order == 1:
        print("Step 2: Solve using separation of variables or integrating factor.")
    elif order == 2:
        print("Step 2: Assume a solution of the form y = e^(rx).")
        print("Step 3: Form the characteristic equation.")
    else:
        print("Step 2: Higher-order equation solved using standard methods.")

    # Analyze solution
    sol_str = str(solution)

    if "exp" in sol_str:
        print("Step 4: Exponential terms appear → typical for linear equations.")

    if "sin" in sol_str or "cos" in sol_str:
        print("Step 5: Sine/cosine terms indicate complex roots.")

    if "C1" in sol_str:
        print("Step 6: Constants (C1, C2, ...) represent general solution.")

    print("Step 7: The solution satisfies the original differential equation.")

except Exception as e:
    print("\n❌ Error:", e)
```
